```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Define the top perpetrator columns explicitly
perp_columns = [
    'Taliban', 'ISIS', 'al-Shabaab', 
    'Houthi extremists (Ansar Allah)', 'Communist Party of India - Maoist (CPI-Maoist)', 
    'ISIS-DRC (formerly ADF)', 'Boko Haram', 
    'Hizballah'
]

# Define the tactic columns
tactic_columns = ['Trauma', 'Assault', 'Covert', 'Coordinated', 'Exploitive']

# Filter the DataFrame to include only the top perpetrators and tactics
subset_df = df_renamed[tactic_columns + ['Perpetrator_Name']]

# One-hot encode the 'Perpetrator_Name' column for correlation analysis
encoded_perp_df = pd.get_dummies(subset_df['Perpetrator_Name'], prefix='', prefix_sep='')

# Combine the encoded perpetrator columns with the tactic columns
combined_df = pd.concat([subset_df[tactic_columns], encoded_perp_df], axis=1)

# Calculate the correlation matrix
correlation_matrix = combined_df.corr()

# Extract the correlation between tactics and top perpetrator columns
perp_tactic_correlation = correlation_matrix.loc[tactic_columns, perp_columns]

# Plot the correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(perp_tactic_correlation, annot=True, cmap='coolwarm', center=0, linecolor='white', linewidths=1, cbar_kws={'shrink': 0.5})
plt.title('Correlation Between Tactics and Top Perpetrator Types')
plt.xlabel('Perpetrator Types')
plt.ylabel('Tactics')
plt.show()
```

```python
from matplotlib.colors import LinearSegmentedColormap
import matplotlib.pyplot as plt
import seaborn as sns

# Create a custom diverging colormap with specified colors for negative (red) and positive (green) correlations
custom_cmap = LinearSegmentedColormap.from_list("custom_red_green", ["#9a031e", "#ffffff", "#0a4719"])

# Plot the correlation matrix with the custom color gradient
plt.figure(figsize=(12, 10))
sns.heatmap(
    perp_tactic_correlation,
    annot=True,
    cmap=custom_cmap,
    center=0,
    linecolor='white',
    linewidths=1,
    cbar_kws={'shrink': 0.5},
    fmt=".2f",
    annot_kws={"fontweight": "bold", "fontname": "Calibri"}
)
plt.title('Correlation Between Tactics and Top Perpetrator Types (Custom Red-Green Scale)')
plt.xlabel('Perpetrator Types')
plt.ylabel('Tactics')
plt.show()
```

```python
# Transpose the correlation matrix to swap x-axis and y-axis
transposed_correlation = perp_tactic_correlation.T

# Plot the transposed correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(
    transposed_correlation,
    annot=True,
    cmap=custom_cmap,
    center=0,
    linecolor='white',
    linewidths=1,
    cbar_kws={'shrink': 0.5},
    fmt=".2f",
    annot_kws={"fontweight": "bold", "fontname": "Calibri"}
)
plt.title('Correlation Between Tactics and Top Perpetrator Types (Swapped Axes)')
plt.xlabel('Tactics')
plt.ylabel('Perpetrator Types')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define the tactic columns
tactic_columns = ['Trauma', 'Assault', 'Covert', 'Coordinated', 'Exploitive']

# Calculate the count of each tactic across the dataset
tactic_counts = df_renamed[tactic_columns].sum()

# Colors for the donut chart using shades of purple
colors = ['#b39ddb', '#9575cd', '#7e57c2', '#673ab7', '#5e35b1']

# Create a donut chart
plt.figure(figsize=(8, 8))
plt.pie(
    tactic_counts.values, 
    labels=tactic_counts.index, 
    autopct='%1.1f%%', 
    startangle=90, 
    colors=colors, 
    wedgeprops={'edgecolor': 'black'}
)
# Add a white circle in the center to make it a donut chart
center_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(center_circle)

# Add a title
plt.title('Distribution of Tactics (Donut Chart)', fontsize=16, fontweight='bold')

# Display the chart
plt.tight_layout()
plt.show()

# Create a radial column chart
from math import pi

# Data for radial column chart
angles = [n / float(len(tactic_counts)) * 2 * pi for n in range(len(tactic_counts))]
angles += angles[:1]  # Close the circle

# Values for the chart
values = list(tactic_counts.values)
values += values[:1]  # Close the circle

# Plot radial column chart
plt.figure(figsize=(10, 8))
ax = plt.subplot(111, polar=True)
plt.xticks(angles[:-1], tactic_counts.index, fontsize=12)
ax.bar(angles[:-1], values, color=colors, alpha=0.7, edgecolor='black', linewidth=1, align='center', width=0.4)
ax.set_yticks([])
plt.title('Distribution of Tactics (Radial Column Chart)', fontsize=16, fontweight='bold')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Enhanced Donut Chart for Better Aesthetics
plt.figure(figsize=(10, 8))

# Use a lighter color palette with gradients
colors = sns.color_palette("Purples", len(tactic_columns))

# Create the donut chart
wedges, texts, autotexts = plt.pie(
    tactic_counts.values, 
    labels=tactic_counts.index, 
    autopct='%1.1f%%', 
    startangle=90, 
    colors=colors, 
    wedgeprops={'edgecolor': 'black', 'linewidth': 1.2}
)

# Add the white circle in the center for the donut effect
center_circle = plt.Circle((0, 0), 0.65, fc='white')
plt.gca().add_artist(center_circle)

# Customize text aesthetics
plt.setp(autotexts, size=12, weight="bold", color="black")
plt.setp(texts, size=13)

# Add a clean title with improved styling
plt.title('Distribution of Tactics', fontsize=18, fontweight='bold', color="#4a4a4a", pad=20)

# Tighten layout for a polished look
plt.tight_layout()

# Display the enhanced chart
plt.show()
```
